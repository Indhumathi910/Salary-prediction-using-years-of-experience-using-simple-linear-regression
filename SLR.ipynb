```python
 import pandas as pd# Import the pandas library and give it the alias 'pd'
```

```python
Dataset = pd.read_csv("Salary_Data.csv")# Read the CSV file named "Salary_Data.csv" into a pandas DataFrame called Dataset
```

```python
Dataset# Display the contents of the Dataset
```

```python
independent = Dataset [['YearsExperience']]# Select the 'YearsExperience' column from Dataset as the independent variable (X)
```

```python
dependent = Dataset [['Salary']]# Select the 'Salary' column from Dataset as the dependent variable (y)
```

```python
from sklearn.model_selection import train_test_split# Import the train_test_split function from scikit-learn
```

```python
train_test_split(independent,dependent,test_size =0.30,random_state=0)# Split the data into training and testing sets
# test_size=0.30 means 30% of the data will be for testing, 70% for training
# random_state=0 ensures the split is reproducible
```

```python
X_train,X_test,y_train,y_test = train_test_split(independent,dependent,test_size =0.30,random_state=0)# Split data into train and test sets (70% train, 30% test)
```

```python
from sklearn.linear_model import LinearRegression# Import Linear Regression model from sklearn
```

```python
regressor=LinearRegression()# Create (initialize) the Linear Regression model
```

```python
regressor.fit(X_train,y_train)# Train the model using training input (X_train)
# and training output (y_train)
```

```python
weight = regressor.coef_# Get model weight (slope)
```

```python
weight
```

```python
bias = regressor.intercept_# Get model bias (intercept)
```

```python
bias
```

```python
y_predict=regressor.predict(X_test)# Predict output for test data
```

```python
from sklearn.metrics import r2_score# Calculate model accuracy (R² score)
```

```python
r_score =r2_score(y_test,y_predict)
```
